# Marketing Campaign Statistical Validation

## Task 4 – Data Storytelling & Statistical Validation

This notebook validates key business findings using statistical hypothesis testing.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import ttest_ind
```

```python
df = pd.read_csv("../data/marketing_campaign.csv", sep="\t")
df.head()
```

```python
df = df.dropna(subset=["Income"])

df["Total_Spending"] = (
    df["MntWines"] +
    df["MntFruits"] +
    df["MntMeatProducts"] +
    df["MntFishProducts"] +
    df["MntSweetProducts"] +
    df["MntGoldProds"]
)

df.head()
```

# Hypothesis Testing

### Null Hypothesis (H0)

There is no significant difference in spending between high-income and low-income customers.

### Alternative Hypothesis (H1)

There is a significant difference in spending between high-income and low-income customers.

```python
median_income = df["Income"].median()

high_income = df[df["Income"] >= median_income]

low_income = df[df["Income"] < median_income]

print("High Income Customers :", len(high_income))
print("Low Income Customers :", len(low_income))
```

```python
t_stat, p_value = ttest_ind(
    high_income["Total_Spending"],
    low_income["Total_Spending"],
    equal_var=False
)

print("T Statistic :", t_stat)
print("P Value :", p_value)
```

```python
if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Income has a statistically significant effect on customer spending.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Income does not have a statistically significant effect on customer spending.")
```
