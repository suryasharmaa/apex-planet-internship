```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/marketing_campaign.csv", sep="\t")

print(df.head())
print(df.shape)
print(df.info())
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("../visuals", exist_ok=True)

# Income Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Income"].dropna(), bins=30)
plt.title("Income Distribution")
plt.savefig("../visuals/income_distribution.png")
plt.close()

# Education Count
plt.figure(figsize=(8,5))
df["Education"].value_counts().plot(kind="bar")
plt.title("Education Count")
plt.savefig("../visuals/education_count.png")
plt.close()

# Marital Status
plt.figure(figsize=(8,5))
df["Marital_Status"].value_counts().plot(kind="bar")
plt.title("Marital Status")
plt.savefig("../visuals/marital_status.png")
plt.close()

# Product Spending
spending = {
    "Wines": df["MntWines"].sum(),
    "Fruits": df["MntFruits"].sum(),
    "Meat": df["MntMeatProducts"].sum(),
    "Fish": df["MntFishProducts"].sum(),
    "Sweets": df["MntSweetProducts"].sum(),
    "Gold": df["MntGoldProds"].sum()
}

plt.figure(figsize=(8,5))
plt.bar(spending.keys(), spending.values())
plt.title("Product Category Spending")
plt.savefig("../visuals/spending_categories.png")
plt.close()

# Correlation Heatmap
numeric_df = df.select_dtypes(include=["int64","float64"])

plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.savefig("../visuals/correlation_heatmap.png")
plt.close()

print("All visualizations saved successfully!")
```

```python
import matplotlib.pyplot as plt

# KPIs
total_customers = len(df)
avg_income = round(df["Income"].mean(), 2)

avg_spending = round(
    (
        df["MntWines"] +
        df["MntFruits"] +
        df["MntMeatProducts"] +
        df["MntFishProducts"] +
        df["MntSweetProducts"] +
        df["MntGoldProds"]
    ).mean(), 2
)

response_rate = round(df["Response"].mean() * 100, 2)

plt.figure(figsize=(12,8))

plt.text(
    0.05, 0.85,
    f"Total Customers\n{total_customers}",
    fontsize=16,
    bbox=dict(facecolor="lightblue")
)

plt.text(
    0.40, 0.85,
    f"Avg Income\n{avg_income}",
    fontsize=16,
    bbox=dict(facecolor="lightgreen")
)

plt.text(
    0.72, 0.85,
    f"Response Rate\n{response_rate}%",
    fontsize=16,
    bbox=dict(facecolor="orange")
)

plt.text(
    0.35, 0.55,
    f"Avg Spending\n{avg_spending}",
    fontsize=18,
    bbox=dict(facecolor="yellow")
)

plt.axis("off")
plt.title(
    "Marketing Campaign Dashboard",
    fontsize=22,
    pad=20
)

plt.savefig(
    "../dashboard/marketing_dashboard.png",
    bbox_inches="tight"
)

plt.close()

print("Dashboard saved successfully!")
```
